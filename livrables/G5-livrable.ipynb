## 1. Identification du problème

Le problème posé par l'ADEME consiste à optimiser des tournées logistiques afin de minimiser l'impact environnemental des déplacements. Dans sa version la plus simple (un seul véhicule devant visiter un ensemble de villes), ce problème est mondialement connu sous le nom de **Problème du Voyageur de Commerce** (ou TSP pour *Traveling Salesman Problem*).

Cependant, la réalité opérationnelle exige une modélisation plus complexe. Nous devons étendre ce modèle de base pour inclure deux contraintes spécifiques :
1.  **Restrictions de passage sur certaines arêtes :** Prise en compte des routes fermées ou inaccessibles.
2.  **Dépendances entre visites :** Instauration de contraintes de précédence (une ville A doit être visitée avant une ville B).

Notre problème s'apparente donc à une variante spécifique du **Problème de Tournées de Véhicules** (VRP - *Vehicle Routing Problem*), et plus précisément au **SOP** (*Sequential Ordering Problem*) ou au **TSP avec contraintes de précédence** (TSP-PC).

## 2. Définition Mathématique du Problème

### 2.1. Formulation du graphe

Nous modélisons le réseau routier par un graphe orienté $G = (V, E)$, où :
* $V = \{0, 1, 2, ..., n\}$ est l'ensemble des sommets. Le sommet $0$ représente le dépôt.
* $E = \{(i, j) \mid i \in V, j \in V, i \neq j\}$ est l'ensemble des arêtes possibles.
* $E_{interdit} \subset E$ est l'ensemble des tronçons routiers bloqués.
* $P$ est l'ensemble des paires $(i, j)$ définissant une précédence (la ville $i$ doit précéder $j$).
* $c_{ij}$ représente le coût (distance ou temps) de traversée de l'arête $(i, j)$.

### 2.2. Variables de décision

Nous utilisons deux types de variables pour tracer la tournée :
1. Une variable de routage binaire :
$$x_{ij} = \begin{cases} 1 & \text{si le véhicule passe directement de } i \text{ à } j \\ 0 & \text{sinon} \end{cases}$$
2. Une variable de position $u_i \in \mathbb{R}^+$ (ou $\mathbb{N}$) représentant la position (ou le temps de passage) du sommet $i$ dans la tournée.

### 2.3. Fonction Objectif

L'objectif est de minimiser le coût total de la tournée :
$$\text{Minimiser } Z = \sum_{i=0}^{n} \sum_{j=0}^{n} c_{ij} x_{ij}$$

### 2.4. Explicitation des Contraintes

Le système doit respecter les équations suivantes :

**1. Visite unique (degré entrant et sortant) :**
Chaque sommet doit être visité une et une seule fois.
$$\sum_{i=0, i \neq j}^{n} x_{ij} = 1 \quad \forall j \in V$$
$$\sum_{j=0, j \neq j}^{n} x_{ij} = 1 \quad \forall i \in V$$

**2. Élimination des sous-tours (Miller-Tucker-Zemlin) :**
Empêche la formation de cycles déconnectés du dépôt principal.
$$u_i - u_j + n x_{ij} \leq n - 1 \quad \forall i, j \in V \setminus \{0\}, i \neq j$$

**3. Restriction de passage :**
Empêche l'utilisation des routes inaccessibles.
$$x_{ij} = 0 \quad \forall (i, j) \in E_{interdit}$$

**4. Contraintes de précédence :**
Garantit qu'une ville dépendante est visitée après sa condition.
$$u_i < u_j \quad \forall (i, j) \in P$$

## 3. Étude de Complexité

Pour choisir la bonne méthode de résolution (méthode exacte vs heuristique), il est fondamental de prouver la complexité théorique de notre problème.

### 3.1. Problème de décision associé
Le problème d'optimisation défini ci-dessus vise à trouver la tournée de coût minimal. Le problème de décision associé peut se formuler ainsi : *Étant donné un graphe $G$, des ensembles $E_{interdit}$ et $P$, et une constante $L$, existe-t-il une tournée valide respectant toutes les contraintes dont le coût total est inférieur ou égal à $L$ ?*

### 3.2. Complexité du cas de base (TSP)
Notre problème est une extension du Problème du Voyageur de Commerce (TSP). Il est prouvé dans la littérature (Karp, 1972) que le problème de décision du TSP est **NP-complet**. La réduction classique se fait à partir du problème du Cycle Hamiltonien, qui est un problème NP-complet bien connu.

### 3.3. Adaptation à notre variante et preuve de NP-Difficulté
Pour démontrer que notre variante (TSP avec précédence et arêtes interdites) est **NP-difficile**, nous pouvons utiliser le principe de restriction. 

Considérons une instance de notre problème où :
* L'ensemble des restrictions est vide ($E_{interdit} = \emptyset$).
* L'ensemble des précédences est vide ($P = \emptyset$).

Dans ce cas particulier, notre problème redevient exactement le problème du TSP standard. Puisque notre problème général contient le TSP comme un cas particulier (sous-problème strict), il est par définition au moins aussi difficile à résoudre que le TSP. 

En conclusion, **notre problème de mobilité avec contraintes est NP-difficile**. Cela justifie pleinement, pour de grandes instances de réseaux routiers, l'utilisation de méthodes de résolution approchées (heuristiques ou méta-heuristiques) en complément des méthodes exactes qui deviendront inopérantes face à l'explosion combinatoire.
